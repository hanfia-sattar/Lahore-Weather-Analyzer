```python
import requests

url = "https://api.open-meteo.com/v1/forecast"
params = {
    "latitude": 31.55,
    "longitude": 74.35,
    "daily": "temperature_2m_max,temperature_2m_min,precipitation_sum,wind_speed_10m_max",
    "past_days": 30,
    "timezone": "Asia/Karachi"
}

response = requests.get(url, params=params)
data = response.json()


dates = data["daily"]["time"]          # the list of date strings
max_temps = data["daily"]["temperature_2m_max"]     # the list of max temperatures
min_temps = data["daily"]["temperature_2m_min"]    # the list of min temperatures
precipitation_sum = data["daily"]["precipitation_sum"]
wind = data["daily"]["wind_speed_10m_max"]

# Print the weather data
print("Lahore Weather — Last 30 Days")
print("-" * 30)

hottest_temp = -100       # start with an impossibly low number
hottest_date = ""
coldest_temp = 100        # start with an impossibly high number
coldest_date = ""
max_rain = -1  # Start with a negative number so any rain amount beats it
rain_date=""

windy_days = [] # List to store (date, wind_speed) for days with wind above 30 km/h

for i in range(len(dates)):
    date = dates[i]
    high = max_temps[i]
    low = min_temps[i]
    rain =precipitation_sum[i]
    wind_record=wind[i]

    print(f"{date} | High: {high}°C | Low: {low}°C")

    if high > hottest_temp:
        hottest_temp = high    # update the record
        hottest_date = date    # remember which date

    if low < coldest_temp:
        coldest_temp = low    # update the record
        coldest_date = date    # remember which date

    if rain > max_rain:
        max_rain=rain
        rain_date=date

    if wind_record > 30:
        windy_days.append((date, wind_record))

print(" ********* WEATHER SUMMARY ********* ")
print(f"\n🔥 Hottest day: {hottest_date} at {hottest_temp}°C")
print(f"\n🧊 Coldest night: {coldest_date} at {coldest_temp}°C")
print(f"\n🌧️ Highest Rain Recorded: {rain_date} at {max_rain}mm")

if windy_days:
    print("\n🍃 Days with wind above 30 km/h:")
    for date, speed in windy_days:
        print(f"    {date} at {speed} km/h")
else:
    print("\n🍃 No days with wind above 30 km/h recorded.")
```
